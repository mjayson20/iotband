# Phase 3 — BoT-IoT Domain Evaluation
**IoT-SecBand | Pipeline position:** UNSW training → frozen MLP → **BoT-IoT evaluation**

**Goal:** Evaluate how well the frozen UNSW-NB15 MLP generalises to NF-BoT-IoT traffic.

**Key constraints enforced:**
- The MLP is **not retrained** — we evaluate generalisation, not fine-tuning
- The scaler is **not refit** on BoT-IoT — locked UNSW-NB15 RobustScaler applied as-is
- The label encoders are **not refit** — locked UNSW-NB15 encoders applied as-is
- Missing features (`service`, `state`) are filled with UNSW-NB15 training mode values

**Phase 2 baseline (to beat or explain):**
- F1=0.7829, Precision=0.7406, Recall=0.8303, FPR=0.1474 @ threshold=0.74

**Red flag:** F1 > 0.95 is suspicious — extreme imbalance (97.7% attack) can inflate metrics.

**Inputs:**
- `data/raw/unsw_nb15/NF-BoT-IoT.parquet`
- `models/mlp_model.pkl` (frozen Phase 2 MLP)
- `models/scaler_params_locked.json`
- `models/label_encoders.json`

**Outputs:**
- `data/processed/botiot_clean.parquet`
- `outputs/metrics/phase3_alignment.json`
- `outputs/metrics/phase3_results.json`
- `outputs/metrics/P3_01_probability_overlap.png`
- `outputs/metrics/P3_02_feature_drift.png`
- `outputs/metrics/P3_03_threshold_sweep.png`

## 7. Interpretation Guide

**Reading the results:**

| Scenario | What it means | Action |
|---|---|---|
| F1 ≈ 0.78 (same as Phase 2) | Good generalisation across datasets | Proceed to Phase 4 (TFLite export) |
| F1 > 0.78 | BoT-IoT traffic is more separable | Check overlap — if overlap is low, result is genuine |
| F1 > 0.95 | **Red flag** — likely inflated by 97.7% attack imbalance | Check confusion matrix, accuracy should NOT be ~97% |
| F1 < 0.65 | Strong distribution shift | Check feature drift plot — high-KS features may need dropping |
| High drift on `sload`/`rate` | NetFlow-derived features differ from UNSW | Expected — document, do not retune without re-running diagnostics |
| High drift on `service`/`state` | All rows have same value (by design) | Expected — these are approximations |

**Before proceeding to Phase 4 (TFLite export):**
1. No red flags triggered
2. F1 is within ±0.10 of Phase 2 baseline (or genuinely better with documented evidence)
3. Probability overlap is documented and understood
4. The `phase3_results.json` is committed to the repo
